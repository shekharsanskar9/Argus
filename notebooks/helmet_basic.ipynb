```python
names = model.names
print("Model classes:", names)

def find_id(*aliases):
    aliases = {a.lower() for a in aliases}
    for i, n in names.items():
        if n.lower().replace("_", " ").strip() in aliases:
            return i
    return None

RIDER     = find_id("rider", "motorcyclist")
NO_HELMET = find_id("without helmet", "no helmet", "wo helmet")
print(f"RIDER={RIDER}  NO_HELMET={NO_HELMET}")
assert RIDER is not None and NO_HELMET is not None, "Check class names printed above"
```

```python
import cv2, yaml, numpy as np
from pathlib import Path

data = yaml.safe_load(open("/content/dataset/data.yaml"))
root = Path(data.get("path") or "/content/dataset")
img_dir = root / data["val"]
lbl_dir = Path(str(img_dir).replace("/images", "/labels"))

def center_in(outer, inner):
    cx, cy = (inner[0] + inner[2]) / 2, (inner[1] + inner[3]) / 2
    return outer[0] <= cx <= outer[2] and outer[1] <= cy <= outer[3]

def iou(a, b):
    x1, y1 = max(a[0], b[0]), max(a[1], b[1])
    x2, y2 = min(a[2], b[2]), min(a[3], b[3])
    inter = max(0, x2 - x1) * max(0, y2 - y1)
    area = lambda r: (r[2] - r[0]) * (r[3] - r[1])
    return inter / (area(a) + area(b) - inter + 1e-9)

def violating_riders(boxes, classes):
    riders = [b for b, c in zip(boxes, classes) if c == RIDER]
    heads  = [b for b, c in zip(boxes, classes) if c == NO_HELMET]
    return [r for r in riders if any(center_in(r, h) for h in heads)]

def load_gt(lbl_path, w, h):
    boxes, classes = [], []
    if lbl_path.exists():
        for line in open(lbl_path):
            parts = line.split()
            if len(parts) < 5:
                continue
            c, cx, cy, bw, bh = int(parts[0]), *map(float, parts[1:5])
            boxes.append([(cx - bw/2) * w, (cy - bh/2) * h, (cx + bw/2) * w, (cy + bh/2) * h])
            classes.append(c)
    return boxes, classes

TP = FP = FN = 0
mistakes = []   # images to look at by hand

imgs = sorted(p for p in img_dir.iterdir() if p.suffix.lower() in {".jpg", ".jpeg", ".png"})
for img_path in imgs:
    img = cv2.imread(str(img_path))
    h, w = img.shape[:2]

    gt_boxes, gt_cls = load_gt(lbl_dir / (img_path.stem + ".txt"), w, h)
    gt_viol = violating_riders(gt_boxes, gt_cls)

    res = model.predict(img, conf=0.35, verbose=False)[0]
    pred_viol = violating_riders(res.boxes.xyxy.cpu().numpy(),
                                 res.boxes.cls.cpu().numpy().astype(int))

    matched = set()
    tp = 0
    for p in pred_viol:
        best = max(range(len(gt_viol)), key=lambda i: iou(p, gt_viol[i]), default=None)
        if best is not None and best not in matched and iou(p, gt_viol[best]) >= 0.5:
            matched.add(best); tp += 1
    fp, fn = len(pred_viol) - tp, len(gt_viol) - tp
    TP += tp; FP += fp; FN += fn
    if fp or fn:
        mistakes.append((img_path, fp, fn))

precision = TP / (TP + FP + 1e-9)
recall    = TP / (TP + FN + 1e-9)
f1        = 2 * precision * recall / (precision + recall + 1e-9)
print(f"Images: {len(imgs)}   True violations: {TP + FN}")
print(f"Correct: {TP}   False alarms: {FP}   Missed: {FN}")
print(f"Precision: {precision:.3f}   Recall: {recall:.3f}   F1: {f1:.3f}")
print(f"{len(mistakes)} images with errors")
```

```python
import cv2
import easyocr

reader = easyocr.Reader(['en'])

img = cv2.imread("/content/plate_crops/plate_final.jpg")

# Create multiple versions
gray = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)

# Version 1: grayscale
gray_img = gray

# Version 2: threshold
_, thresh = cv2.threshold(
    gray, 0, 255,
    cv2.THRESH_BINARY + cv2.THRESH_OTSU
)

# Version 3: adaptive threshold
adaptive = cv2.adaptiveThreshold(
    gray,
    255,
    cv2.ADAPTIVE_THRESH_GAUSSIAN_C,
    cv2.THRESH_BINARY,
    11,
    2
)

images = {
    "original": img,
    "gray": gray_img,
    "threshold": thresh,
    "adaptive": adaptive
}

best_text = ""
best_conf = 0

for name, image in images.items():

    result = reader.readtext(
        image,
        detail=1,
        allowlist="ABCDEFGHIJKLMNOPQRSTUVWXYZ0123456789"
    )

    for detection in result:
        text = detection[1]
        conf = detection[2]

        print(f"{name}: {text} -> {conf:.4f}")

        if conf > best_conf:
            best_conf = conf
            best_text = text

print("\n======================")
print("BEST OCR RESULT:", best_text)
print("BEST CONFIDENCE:", best_conf)
print("======================")
```

```python
import cv2
import numpy as np

img = cv2.imread("/content/plate_crops/plate_0.jpg")

gray = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)

# Reduce noise
blur = cv2.GaussianBlur(gray, (5, 5), 0)

# Find edges
edges = cv2.Canny(blur, 50, 150)

# Find contours
contours, _ = cv2.findContours(
    edges,
    cv2.RETR_EXTERNAL,
    cv2.CHAIN_APPROX_SIMPLE
)

plate_contour = None

for contour in sorted(contours, key=cv2.contourArea, reverse=True):
    perimeter = cv2.arcLength(contour, True)
    approx = cv2.approxPolyDP(
        contour,
        0.03 * perimeter,
        True
    )

    if len(approx) == 4:
        x, y, w, h = cv2.boundingRect(approx)

        # License plate should be wider than tall
        if w > h * 1.5:
            plate_contour = approx
            break

if plate_contour is None:
    print("Plate contour not found.")
else:
    points = plate_contour.reshape(4, 2).astype(np.float32)

    # Sort corners
    s = points.sum(axis=1)
    d = np.diff(points, axis=1).flatten()

    top_left = points[np.argmin(s)]
    bottom_right = points[np.argmax(s)]
    top_right = points[np.argmin(d)]
    bottom_left = points[np.argmax(d)]

    src = np.array([
        top_left,
        top_right,
        bottom_right,
        bottom_left
    ], dtype=np.float32)

    width = 400
    height = 120

    dst = np.array([
        [0, 0],
        [width - 1, 0],
        [width - 1, height - 1],
        [0, height - 1]
    ], dtype=np.float32)

    matrix = cv2.getPerspectiveTransform(src, dst)

    corrected = cv2.warpPerspective(
        img,
        matrix,
        (width, height)
    )

    cv2.imwrite(
        "/content/plate_crops/plate_corrected.jpg",
        corrected
    )

    print("Perspective correction completed.")
```

```python
import cv2
import os
import pandas as pd

os.makedirs("/content/plate_results", exist_ok=True)

results_table = []

for image_name in os.listdir("/content/dataset/val/images"):

    if not image_name.lower().endswith((".jpg", ".jpeg", ".png")):
        continue

    image_path = f"/content/dataset/val/images/{image_name}"
    image = cv2.imread(image_path)

    # YOLO detection
    result = model(image_path)[0]

    plate_count = 0

    for box, cls, conf in zip(
        result.boxes.xyxy,
        result.boxes.cls,
        result.boxes.conf
    ):

        # Class 3 = number plate
        if int(cls) != 3:
            continue

        x1, y1, x2, y2 = map(int, box)

        # Make sure coordinates are inside image
        h, w = image.shape[:2]
        x1 = max(0, x1)
        y1 = max(0, y1)
        x2 = min(w, x2)
        y2 = min(h, y2)

        plate = image[y1:y2, x1:x2]

        if plate.size == 0:
            continue

        # Upscale
        plate = cv2.resize(
            plate,
            None,
            fx=5,
            fy=5,
            interpolation=cv2.INTER_CUBIC
        )

        crop_name = f"{os.path.splitext(image_name)[0]}_{plate_count}.jpg"
        crop_path = f"/content/plate_results/{crop_name}"

        cv2.imwrite(crop_path, plate)

        # OCR
        ocr_result = reader.readtext(
            plate,
            detail=1,
            allowlist="ABCDEFGHIJKLMNOPQRSTUVWXYZ0123456789"
        )

        if ocr_result:
            best = max(ocr_result, key=lambda x: x[2])
            text = best[1]
            ocr_conf = float(best[2])
        else:
            text = ""
            ocr_conf = 0.0

        results_table.append({
            "image": image_name,
            "plate": text,
            "ocr_confidence": round(ocr_conf, 4),
            "yolo_confidence": round(float(conf), 4),
            "crop": crop_path
        })

        plate_count += 1

# Create results table
df = pd.DataFrame(results_table)

print("Total plates detected:", len(df))
display(df)
```

```python
import cv2
import os
import re
import pandas as pd

def clean_text(text):
    return re.sub(r'[^A-Z0-9]', '', text.upper())


def run_ocr(plate):
    # Resize
    plate = cv2.resize(
        plate,
        None,
        fx=5,
        fy=5,
        interpolation=cv2.INTER_CUBIC
    )

    gray = cv2.cvtColor(plate, cv2.COLOR_BGR2GRAY)

    # Different preprocessing versions
    images = [
        plate,
        gray,
        cv2.threshold(
            gray, 0, 255,
            cv2.THRESH_BINARY + cv2.THRESH_OTSU
        )[1]
    ]

    candidates = []

    for img in images:

        result = reader.readtext(
            img,
            detail=1,
            allowlist="ABCDEFGHIJKLMNOPQRSTUVWXYZ0123456789"
        )

        for _, text, conf in result:
            text = clean_text(text)

            if text:
                candidates.append((text, float(conf)))

    if not candidates:
        return "", 0.0

    # Prefer longer valid text, while considering confidence
    candidates.sort(
        key=lambda x: (len(x[0]) >= 6, x[1], len(x[0])),
        reverse=True
    )

    return candidates[0]


results = []

for filename in os.listdir("/content/plate_results"):

    if not filename.endswith(".jpg"):
        continue

    path = f"/content/plate_results/{filename}"

    plate = cv2.imread(path)

    text, confidence = run_ocr(plate)

    results.append({
        "image": filename,
        "plate": text,
        "confidence": round(confidence, 4)
    })

df = pd.DataFrame(results)

display(df)
```

```python
import cv2
import matplotlib.pyplot as plt

img = cv2.imread("/content/plate_results/new0_0.jpg")

gray = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)

# Detect edges
edges = cv2.Canny(gray, 50, 150)

# Detect lines
lines = cv2.HoughLinesP(
    edges,
    1,
    3.14159 / 180,
    threshold=50,
    minLineLength=50,
    maxLineGap=10
)

angles = []

if lines is not None:
    for line in lines:
        x1, y1, x2, y2 = line[0]

        angle = __import__("math").degrees(
            __import__("math").atan2(y2 - y1, x2 - x1)
        )

        # Only consider reasonably horizontal plate lines
        if abs(angle) < 45:
            angles.append(angle)

print("Detected angles:", angles)

if angles:
    angle = sum(angles) / len(angles)
else:
    angle = 0

print("Rotation angle:", angle)

# Rotate
h, w = img.shape[:2]
center = (w // 2, h // 2)

M = cv2.getRotationMatrix2D(center, angle, 1.0)

rotated = cv2.warpAffine(
    img,
    M,
    (w, h),
    borderMode=cv2.BORDER_REPLICATE
)

plt.figure(figsize=(12, 5))
plt.imshow(cv2.cvtColor(rotated, cv2.COLOR_BGR2RGB))
plt.axis("off");
```

```python
import cv2
import os
import numpy as np
import pandas as pd
import re
import math

def get_angle(img):
    gray = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)
    edges = cv2.Canny(gray, 50, 150)

    lines = cv2.HoughLinesP(
        edges,
        1,
        np.pi / 180,
        threshold=50,
        minLineLength=50,
        maxLineGap=10
    )

    if lines is None:
        return 0

    angles = []

    for line in lines:
        x1, y1, x2, y2 = line[0]

        angle = math.degrees(
            math.atan2(y2 - y1, x2 - x1)
        )

        # Plate is approximately horizontal
        if -45 < angle < 45:
            angles.append(angle)

    if not angles:
        return 0

    return np.median(angles)


def process_plate(path):

    img = cv2.imread(path)

    # 1. Find plate angle
    angle = get_angle(img)

    # 2. Rotate to straighten plate
    h, w = img.shape[:2]
    center = (w // 2, h // 2)

    M = cv2.getRotationMatrix2D(
        center,
        angle,
        1.0
    )

    rotated = cv2.warpAffine(
        img,
        M,
        (w, h),
        borderMode=cv2.BORDER_REPLICATE
    )

    # 3. OCR
    result = reader.readtext(
        rotated,
        detail=1,
        allowlist="ABCDEFGHIJKLMNOPQRSTUVWXYZ0123456789"
    )

    candidates = []

    for box, text, confidence in result:

        text = re.sub(
            r'[^A-Z0-9]',
            '',
            text.upper()
        )

        # Ignore short detections such as YAHHA, H, etc.
        if len(text) >= 6:
            candidates.append(
                (text, float(confidence))
            )

    # 4. Select best candidate
    if candidates:

        candidates.sort(
            key=lambda x: (len(x[0]), x[1]),
            reverse=True
        )

        best_text, best_conf = candidates[0]

    else:
        best_text = ""
        best_conf = 0.0

    return best_text, best_conf, angle


results = []

for filename in sorted(os.listdir("/content/plate_results")):

    if not filename.lower().endswith(".jpg"):
        continue

    path = f"/content/plate_results/{filename}"

    text, confidence, angle = process_plate(path)

    results.append({
        "image": filename,
        "plate": text,
        "confidence": round(confidence, 4),
        "angle": round(angle, 2)
    })


df_final = pd.DataFrame(results)

display(df_final)
```

```python
import re
import pandas as pd

def normalize_ocr(text):
    text = text.upper()
    text = re.sub(r'[^A-Z0-9]', '', text)

    # Common OCR confusions
    replacements = {
        'O': '0',
        'I': '1',
        'L': '1'
    }

    return ''.join(replacements.get(c, c) for c in text)


def plate_score(text, confidence):
    """
    Score an OCR candidate based on:
    - length
    - Indian-style plate structure
    - OCR confidence
    """

    text = normalize_ocr(text)

    if len(text) < 6:
        return -1

    score = 0

    # Normal Indian plate pattern:
    # XX00XX0000
    pattern1 = r'^[A-Z]{2}[0-9]{1,2}[A-Z]{1,3}[0-9]{1,4}$'

    # Allow OCR to miss the state letters slightly
    pattern2 = r'^[A-Z]{1,3}[0-9]{1,2}[A-Z]{1,3}[0-9]{1,4}$'

    if re.match(pattern1, text):
        score += 100

    elif re.match(pattern2, text):
        score += 60

    # Reasonable plate length
    if 8 <= len(text) <= 12:
        score += 30
    elif 6 <= len(text) <= 13:
        score += 10

    # OCR confidence
    score += confidence * 20

    return score


def get_best_plate(candidates):

    scored = []

    for candidate in candidates:

        text = candidate["text"]
        confidence = candidate["confidence"]

        score = plate_score(text, confidence)

        if score >= 0:
            scored.append({
                "text": text,
                "confidence": confidence,
                "score": score,
                "method": candidate["method"]
            })

    if not scored:
        return "", 0.0, ""

    best = max(scored, key=lambda x: x["score"])

    return (
        best["text"],
        best["confidence"],
        best["method"]
    )


final_results = []

for filename in sorted(os.listdir("/content/plate_results")):

    if not filename.lower().endswith(".jpg"):
        continue

    path = f"/content/plate_results/{filename}"

    candidates = ocr_variants(path)

    plate, confidence, method = get_best_plate(candidates)

    final_results.append({
        "image": filename,
        "plate": plate,
        "confidence": round(confidence, 4),
        "method": method
    })


df_final = pd.DataFrame(final_results)

display(df_final)
```

```python
import cv2
import os
import re
import numpy as np
import pandas as pd

# ---------------------------------------------------------
# 1. OCR candidate generation
# ---------------------------------------------------------

def clean_text(text):
    return re.sub(r'[^A-Z0-9]', '', text.upper())


def generate_ocr_candidates(image):
    # accepts a file path or a BGR image array

    img = cv2.imread(image) if isinstance(image, str) else image

    if img is None:
        return []

    # Upscale
    img = cv2.resize(
        img,
        None,
        fx=4,
        fy=4,
        interpolation=cv2.INTER_CUBIC
    )

    gray = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)

    # Mild sharpening
    kernel = np.array([
        [0, -1, 0],
        [-1, 5, -1],
        [0, -1, 0]
    ])

    sharpened = cv2.filter2D(gray, -1, kernel)

    # OTSU
    otsu = cv2.threshold(
        gray,
        0,
        255,
        cv2.THRESH_BINARY + cv2.THRESH_OTSU
    )[1]

    # Adaptive threshold
    adaptive = cv2.adaptiveThreshold(
        gray,
        255,
        cv2.ADAPTIVE_THRESH_GAUSSIAN_C,
        cv2.THRESH_BINARY,
        31,
        11
    )

    variants = {
        "original": img,
        "gray": gray,
        "sharp": sharpened,
        "otsu": otsu,
        "adaptive": adaptive
    }

    candidates = []

    for method, processed in variants.items():

        result = reader.readtext(
            processed,
            detail=1,
            allowlist="ABCDEFGHIJKLMNOPQRSTUVWXYZ0123456789"
        )

        for box, text, confidence in result:

            text = clean_text(text)

            if not text:
                continue

            candidates.append({
                "text": text,
                "confidence": float(confidence),
                "method": method
            })

    return candidates


# ---------------------------------------------------------
# 2. Rank OCR candidates
# ---------------------------------------------------------

def candidate_score(text, confidence):

    length = len(text)

    score = confidence * 100

    # License plates generally contain several characters
    if length >= 8:
        score += 20
    elif length >= 6:
        score += 10
    elif length <= 3:
        score -= 30

    # Penalize obvious motorcycle/company words
    bad_words = [
        "YAMAHA",
        "HONDA",
        "HERO",
        "BAJAJ",
        "TVS",
        "ROYAL"
    ]

    if text in bad_words:
        score -= 80

    return score


# ---------------------------------------------------------
# 3. Process all existing plate crops
# ---------------------------------------------------------

final_results = []

plate_dir = "/content/plate_results"

for filename in sorted(os.listdir(plate_dir)):

    if not filename.lower().endswith(
        (".jpg", ".jpeg", ".png")
    ):
        continue

    path = os.path.join(plate_dir, filename)

    candidates = generate_ocr_candidates(path)

    if not candidates:

        final_results.append({
            "image": filename,
            "plate": "",
            "confidence": 0,
            "method": "",
            "num_candidates": 0
        })

        continue

    # Rank candidates
    ranked = sorted(
        candidates,
        key=lambda x: candidate_score(
            x["text"],
            x["confidence"]
        ),
        reverse=True
    )

    best = ranked[0]

    final_results.append({
        "image": filename,
        "plate": best["text"],
        "confidence": round(best["confidence"], 4),
        "method": best["method"],
        "num_candidates": len(candidates)
    })


# ---------------------------------------------------------
# 4. Create final dataframe
# ---------------------------------------------------------

df_final = pd.DataFrame(final_results)

display(df_final)


# ---------------------------------------------------------
# 5. Save results
# ---------------------------------------------------------

output_path = "/content/final_ocr_results.csv"

df_final.to_csv(
    output_path,
    index=False
)

print()
print("Saved:", output_path)
```

```python
import cv2, os, numpy as np, pandas as pd

PLATE = find_id("number plate", "license plate", "licence plate", "plate")
assert PLATE is not None, "Check class names printed earlier"

def box_center(b):
    return ((b[0] + b[2]) / 2, (b[1] + b[3]) / 2)

def plate_for_rider(rider, plates):
    """Plate whose centre is inside the rider box, else the nearest plate."""
    if not plates:
        return None
    inside = [p for p in plates if center_in(rider, p)]
    if inside:
        return inside[0]
    rx, ry = box_center(rider)
    return min(plates, key=lambda p: (box_center(p)[0] - rx) ** 2 + (box_center(p)[1] - ry) ** 2)

def read_plate(crop):
    """Best OCR reading for a plate crop as (text, confidence)."""
    candidates = generate_ocr_candidates(crop)
    if not candidates:
        return "", 0.0
    best = max(candidates, key=lambda c: candidate_score(c["text"], c["confidence"]))
    return best["text"], best["confidence"]

def crop_box(img, box, pad=0.1):
    h, w = img.shape[:2]
    x1, y1, x2, y2 = box
    px, py = (x2 - x1) * pad, (y2 - y1) * pad
    x1, y1 = int(max(0, x1 - px)), int(max(0, y1 - py))
    x2, y2 = int(min(w, x2 + px)), int(min(h, y2 + py))
    return img[y1:y2, x1:x2]

def analyse_image(img, conf=0.35):
    """Return the violations found in one BGR image, each with its plate reading."""
    res = model.predict(img, conf=conf, verbose=False)[0]
    boxes = res.boxes.xyxy.cpu().numpy()
    classes = res.boxes.cls.cpu().numpy().astype(int)
    plates = [b for b, c in zip(boxes, classes) if c == PLATE]

    violations = []
    for rider in violating_riders(boxes, classes):
        plate_box = plate_for_rider(rider, plates)
        text, ocr_conf = "", 0.0
        if plate_box is not None:
            crop = crop_box(img, plate_box)
            if crop.size:
                text, ocr_conf = read_plate(crop)
        violations.append({"rider": rider, "plate_box": plate_box,
                           "plate": text, "ocr_confidence": ocr_conf})
    return violations

def draw_violations(img, violations):
    out = img.copy()
    for v in violations:
        x1, y1, x2, y2 = map(int, v["rider"])
        cv2.rectangle(out, (x1, y1), (x2, y2), (0, 0, 255), 2)
        if v["plate_box"] is not None:
            px1, py1, px2, py2 = map(int, v["plate_box"])
            cv2.rectangle(out, (px1, py1), (px2, py2), (0, 165, 255), 2)
        label = "NO HELMET " + (v["plate"] or "plate unreadable")
        cv2.putText(out, label, (x1, max(15, y1 - 6)),
                    cv2.FONT_HERSHEY_SIMPLEX, 0.5, (0, 0, 255), 2)
    return out
```

```python
# --- Load model from Drive + define every helper the video test needs ---
from google.colab import drive
from ultralytics import YOLO
import easyocr, cv2, os, re, numpy as np, pandas as pd

drive.mount('/content/drive')
MODEL_PATH = '/content/drive/MyDrive/Helmet_Violation_Detection/helmet_plate_yolov8_best.pt'  # <-- change if needed
assert os.path.exists(MODEL_PATH), f'Model not found: {MODEL_PATH}'
model = YOLO(MODEL_PATH)
reader = easyocr.Reader(['en'])
print('Model classes:', model.names)

def find_id(*aliases):
    aliases = {a.lower() for a in aliases}
    for i, n in model.names.items():
        if n.lower().replace('_', ' ').strip() in aliases:
            return i
    return None

RIDER     = find_id('rider', 'motorcyclist')
NO_HELMET = find_id('without helmet', 'no helmet', 'wo helmet')
PLATE     = find_id('number plate', 'license plate', 'licence plate', 'plate')
assert None not in (RIDER, NO_HELMET, PLATE), f'Check class names: {model.names}'
print(f'RIDER={RIDER}  NO_HELMET={NO_HELMET}  PLATE={PLATE}')

def center_in(outer, inner):
    cx, cy = (inner[0]+inner[2])/2, (inner[1]+inner[3])/2
    return outer[0] <= cx <= outer[2] and outer[1] <= cy <= outer[3]

def box_center(b):
    return ((b[0]+b[2])/2, (b[1]+b[3])/2)

def plate_for_rider(rider, plates):
    if not plates:
        return None
    inside = [p for p in plates if center_in(rider, p)]
    if inside:
        return inside[0]
    rx, ry = box_center(rider)
    return min(plates, key=lambda p: (box_center(p)[0]-rx)**2 + (box_center(p)[1]-ry)**2)

def crop_box(img, box, pad=0.1):
    h, w = img.shape[:2]
    x1, y1, x2, y2 = box
    px, py = (x2-x1)*pad, (y2-y1)*pad
    x1, y1 = int(max(0, x1-px)), int(max(0, y1-py))
    x2, y2 = int(min(w, x2+px)), int(min(h, y2+py))
    return img[y1:y2, x1:x2]

def clean_text(t):
    return re.sub(r'[^A-Z0-9]', '', t.upper())

def generate_ocr_candidates(image):
    img = cv2.imread(image) if isinstance(image, str) else image
    if img is None:
        return []
    img = cv2.resize(img, None, fx=4, fy=4, interpolation=cv2.INTER_CUBIC)
    gray = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)
    sharp = cv2.filter2D(gray, -1, np.array([[0,-1,0],[-1,5,-1],[0,-1,0]]))
    otsu = cv2.threshold(gray, 0, 255, cv2.THRESH_BINARY+cv2.THRESH_OTSU)[1]
    adaptive = cv2.adaptiveThreshold(gray, 255, cv2.ADAPTIVE_THRESH_GAUSSIAN_C, cv2.THRESH_BINARY, 31, 11)
    variants = {'original': img, 'gray': gray, 'sharp': sharp, 'otsu': otsu, 'adaptive': adaptive}
    out = []
    for m, proc in variants.items():
        for box, text, conf in reader.readtext(proc, detail=1, allowlist='ABCDEFGHIJKLMNOPQRSTUVWXYZ0123456789'):
            text = clean_text(text)
            if text:
                out.append({'text': text, 'confidence': float(conf), 'method': m})
    return out

def candidate_score(text, conf):
    L = len(text)
    s = conf*100
    if L >= 8:
        s += 20
    elif L >= 6:
        s += 10
    elif L <= 3:
        s -= 30
    if text in ['YAMAHA','HONDA','HERO','BAJAJ','TVS','ROYAL']:
        s -= 80
    return s

def read_plate(crop):
    c = generate_ocr_candidates(crop)
    if not c:
        return '', 0.0
    best = max(c, key=lambda x: candidate_score(x['text'], x['confidence']))
    return best['text'], best['confidence']

def draw_violations(img, violations):
    out = img.copy()
    for v in violations:
        x1, y1, x2, y2 = map(int, v['rider'])
        cv2.rectangle(out, (x1, y1), (x2, y2), (0, 0, 255), 2)
        if v.get('plate_box') is not None:
            px1, py1, px2, py2 = map(int, v['plate_box'])
            cv2.rectangle(out, (px1, py1), (px2, py2), (0, 165, 255), 2)
        plate = v.get('plate')
        label = 'NO HELMET | ' + (plate if plate else 'NO PLATE')
        cv2.putText(out, label, (x1, max(15, y1-6)), cv2.FONT_HERSHEY_SIMPLEX, 0.6, (0, 0, 255), 2)
    return out

def process_video(video_path, out_path, conf=0.3):
    cap = cv2.VideoCapture(video_path)
    assert cap.isOpened(), f'Cannot open {video_path}'
    fps = cap.get(cv2.CAP_PROP_FPS) or 25
    w = int(cap.get(cv2.CAP_PROP_FRAME_WIDTH))
    h = int(cap.get(cv2.CAP_PROP_FRAME_HEIGHT))
    writer = cv2.VideoWriter(out_path, cv2.VideoWriter_fourcc(*'mp4v'), fps, (w, h))
    seen = {}
    while True:
        ok, frame = cap.read()
        if not ok:
            break
        res = model.track(frame, conf=conf, persist=True, verbose=False)[0]
        if res.boxes.id is None:
            writer.write(frame)
            continue
        boxes = res.boxes.xyxy.cpu().numpy()
        classes = res.boxes.cls.cpu().numpy().astype(int)
        ids = res.boxes.id.cpu().numpy().astype(int)
        plates = [b for b, c in zip(boxes, classes) if c == PLATE]
        violations = []
        for rider, tid, c in zip(boxes, ids, classes):
            if c != RIDER or not any(center_in(rider, hd) for hd, k in zip(boxes, classes) if k == NO_HELMET):
                continue
            plate_box = plate_for_rider(rider, plates)
            best = seen.setdefault(tid, {'plate': '', 'ocr_confidence': 0.0})
            if plate_box is not None:
                crop = crop_box(frame, plate_box)
                if crop.size:
                    text, oc = read_plate(crop)
                    if oc > best['ocr_confidence']:
                        best.update(plate=text, ocr_confidence=oc)
            violations.append({'rider': rider, 'plate_box': plate_box, 'plate': best['plate']})
        writer.write(draw_violations(frame, violations))
    cap.release()
    writer.release()
    return pd.DataFrame([{'track_id': t, **v} for t, v in seen.items()])

print('✅ Model + helpers + process_video ready.')
```

```python
# --- FAST + COMPLETE run: 720p, frame-skip, progress, H.264 output ---
import os, re
OUT_RAW  = '/content/annotated_raw.mp4'
OUT_H264 = '/content/annotated_output.mp4'
CROP_DIR = '/content/plate_evidence'; os.makedirs(CROP_DIR, exist_ok=True)
RIDER_CONF, PLATE_CONF, MIN_CONF, FRAME_SKIP = 0.25, 0.10, 0.30, 2

def plate_quality(text, conf):
    t = re.sub(r'[^A-Z0-9]', '', text.upper())
    if not t:
        return -1, t
    s = conf*40 + min(len(t), 10)*8
    if re.match(r'^[A-Z]{2}[0-9]{1,2}[A-Z]{0,3}[0-9]{3,4}$', t):
        s += 60
    if t in ['YAMAHA', 'HONDA', 'HERO', 'BAJAJ', 'TVS', 'ROYAL']:
        s -= 100
    return s, t

def read_plate_joined(crop):
    h = crop.shape[0]
    if h < 96:
        f = 96.0/max(h, 1); crop = cv2.resize(crop, None, fx=f, fy=f, interpolation=cv2.INTER_CUBIC)
    big  = cv2.resize(crop, None, fx=3, fy=3, interpolation=cv2.INTER_CUBIC)
    gray = cv2.cvtColor(big, cv2.COLOR_BGR2GRAY)
    best = (-1, '', 0.0)
    for img in (big, gray):
        results = reader.readtext(img, detail=1, allowlist='ABCDEFGHIJKLMNOPQRSTUVWXYZ0123456789')
        if not results:
            continue
        results = sorted(results, key=lambda r: r[0][0][0])
        text = ''.join(r[1] for r in results)
        conf = float(np.mean([r[2] for r in results]))
        s, t = plate_quality(text, conf)
        if s > best[0]:
            best = (s, t, conf)
    return best

cap = cv2.VideoCapture(VIDEO_PATH)
assert cap.isOpened(), f'Cannot open {VIDEO_PATH}'
total = int(cap.get(cv2.CAP_PROP_FRAME_COUNT)) or 0
fps = (cap.get(cv2.CAP_PROP_FPS) or 25) / FRAME_SKIP
OW, OH = 1280, 720
writer = cv2.VideoWriter(OUT_RAW, cv2.VideoWriter_fourcc(*'mp4v'), fps, (OW, OH))

seen = {}; counts = {'frames': 0, 'plates_detected': 0, 'plates_read': 0}; fi = 0
while True:
    ok, frame = cap.read()
    if not ok:
        break
    fi += 1
    if fi % FRAME_SKIP:
        continue
    frame = cv2.resize(frame, (OW, OH)); counts['frames'] += 1
    if counts['frames'] % 100 == 0:
        print(f"  ...{counts['frames']} frames done (src {fi}/{total})")
    res  = model.track(frame, conf=RIDER_CONF, persist=True, verbose=False)[0]
    pres = model.predict(frame, conf=PLATE_CONF, verbose=False)[0]
    pcls = pres.boxes.cls.cpu().numpy().astype(int); pbox = pres.boxes.xyxy.cpu().numpy()
    plates = [b for b, c in zip(pbox, pcls) if c == PLATE]; counts['plates_detected'] += len(plates)
    if res.boxes.id is None:
        writer.write(frame); continue
    boxes = res.boxes.xyxy.cpu().numpy(); classes = res.boxes.cls.cpu().numpy().astype(int)
    ids = res.boxes.id.cpu().numpy().astype(int); violations = []
    for rider, tid, c in zip(boxes, ids, classes):
        if c != RIDER or not any(center_in(rider, hd) for hd, k in zip(boxes, classes) if k == NO_HELMET):
            continue
        plate_box = plate_for_rider(rider, plates)
        best = seen.setdefault(tid, {'plate': '', 'score': -1, 'conf': 0.0})
        if plate_box is not None:
            crop = crop_box(frame, plate_box, pad=0.2)
            if crop.size:
                score, text, conf = read_plate_joined(crop)
                if text:
                    counts['plates_read'] += 1
                if score > best['score']:
                    best.update(plate=text, score=score, conf=conf)
                    cv2.imwrite(f'{CROP_DIR}/rider_{tid}.jpg', crop)
        shown = best['plate'] if best['conf'] >= MIN_CONF else ''
        violations.append({'rider': rider, 'plate_box': plate_box, 'plate': shown})
    writer.write(draw_violations(frame, violations))
cap.release(); writer.release()        # finalises the file (writes the moov atom)
print("Video written. Re-encoding to H.264...")

os.system(f'ffmpeg -y -i {OUT_RAW} -vcodec libx264 -pix_fmt yuv420p {OUT_H264}')

rows = []
for t, v in seen.items():
    readable = v['conf'] >= MIN_CONF and v['plate']
    rows.append({'track_id': t, 'plate': v['plate'] if readable else 'NO PLATE',
                 'ocr_confidence': round(v['conf'], 3), 'status': 'readable' if readable else 'NO PLATE'})
df = pd.DataFrame(rows); df.to_csv('/content/video_violations.csv', index=False)
readable = (df['status'] == 'readable').sum()
print("DIAGNOSTIC:", counts)
print(f"{len(df)} violating riders | {readable} with plate | {len(df)-readable} NO PLATE")
display(df)

from google.colab import files
files.download(OUT_H264)
```
